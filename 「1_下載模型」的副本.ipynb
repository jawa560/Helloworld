#此腳本用於將LLM模型離線下載到雲端硬碟中
注意:雲端硬碟空間至少需15G左右

1.安裝huggingface並登入

請先取得huggingface_hub Access Token

網址：https://huggingface.co/

```python
# 安裝 Hugging Face CLI 工具
!pip install -q huggingface_hub

# 登入 Hugging Face（只需執行一次，會跳出輸入 Access Token 的提示）
from huggingface_hub import notebook_login
notebook_login()
```

2.掛載 Google Drive

將模型下載後存入自己的雲端硬碟，這樣以後使用都不需要下載

```python
from google.colab import drive
drive.mount('/content/drive')
```

3.指定你的存放路徑

可以先從左側資料夾建立好預期路徑後，點選複製路徑，貼到下面

```python
model_path = "/content/drive/MyDrive/LLMTEST/LLaVA-1.5-7B-OFFLINE" #路徑
```

4.下載離線模型至指定路徑，5~10分鐘

```python
from huggingface_hub import snapshot_download

# 下載完整模型（大約 13GB）
snapshot_download(
    repo_id="llava-hf/llava-1.5-7b-hf", #模型名稱
    local_dir=model_path,
    local_dir_use_symlinks=False
)
```
